Question 3: Which drivers are the most consistent in their lap times? To measure driver consistency, we can examine the variation in lap times across races per circuit. A more consistent driver will have lower variability in their lap times.

Relevant tables: drivers, races, lapTimes, circuits

```python
# Import relevant data
circuits_df = pd.read_csv("../../data/ergast/circuits.csv")
drivers_df = pd.read_csv("../../data/ergast/drivers.csv")
races_df = pd.read_csv("../../data/ergast/races.csv")
lap_times_df = pd.read_csv("../../data/ergast/lap_times.csv")
```

```python
# Check for null values and sum them per column
circuits_null_counts_per_column = circuits_df.isnull().sum()
print("circuits_df Null counts per column:\n", circuits_null_counts_per_column)

# Check for data types 
print(circuits_df.dtypes)

# Drop unncessary columns
circuits_df = circuits_df.drop(['circuitRef','location', 'lat', 'lng', 'alt', 'url'], axis=1)

# Rename column
circuits_df.rename(columns={'name': 'circuit_name'}, inplace=True)

print("Final columns: ", circuits_df.columns)
```

```python
# Check for null values and sum them per column
drivers_null_counts_per_column = drivers_df.isnull().sum()
print("drivers_df Null counts per column:\n", drivers_null_counts_per_column)

# Check for data types 
print(drivers_df.dtypes)

# Keep necessary columns
drivers_df = drivers_df[['driverId', 'driverRef']]


print("Final columns: ", drivers_df.columns)
```

```python
# Check for null values and sum them per column
races_null_counts_per_column = races_df.isnull().sum()
print("races_df Null counts per column:\n", races_null_counts_per_column, "\n")

# Check for data types 
print(races_df.dtypes)

# Convert date column to datetime object
races_df['date'] = pd.to_datetime(races_df['date'])

# Filter column value so only include 2009 and onwards
races_df = races_df[races_df['year'] >= 2009 ]

# Drop unnecessary columns
races_df = races_df.drop(['time', 'url', 'fp1_date', 'fp1_time', 
'fp2_date', 'fp2_time', 'fp3_date', 'fp3_time', 'quali_date', 'quali_time', 'sprint_date', 'sprint_time'], axis = 1)

# Rename column
races_df.rename(columns={'name': 'race_name'}, inplace = True)

print("Final columns: ", races_df.columns)
```

```python
# Check for null values and sum them per column
lap_times_null_counts_per_column = lap_times_df.isnull().sum()
print("lap_times Null counts per column:\n", lap_times_null_counts_per_column, "\n")

# Check for data types 
print(lap_times_df.dtypes)

# Drop column
lap_times_df.drop(['position'], axis = 1, inplace=True)

# Rename
lap_times_df.rename(columns = {'lap': 'lap_number', 'time': 'lap_time', 'milliseconds': 'lap_millisec'}, inplace=True)

print("Final columns: ", lap_times_df.columns)
```

Now we merge these dataframes and create a new dataframe to address this question.

```python
# Merge lap times with drivers, races, and circuits
consistency_df = lap_times_df.merge(drivers_df, on='driverId', how='left')
consistency_df = consistency_df.merge(races_df[['raceId', 'circuitId']], on='raceId', how='left')
consistency_df = consistency_df.merge(circuits_df[['circuitId', 'circuit_name']], on='circuitId', how='left')

# Keep only relevant columns for plotting
consistency_df = consistency_df[['driverId', 'driverRef', 'circuitId', 'circuit_name', 'lap_millisec']]

# Sort for clarity
consistency_df = consistency_df.sort_values(by=['driverId', 'circuitId'])

print(consistency_df)
```

```python
consistency_df.to_csv('../../data/lap_times_per_driver_and_circuit.csv', index=False)
```

Question 4: How do lap times evolve throughout a race? To measure lap time variability, we can observe whether lap times generally decrease or increase, and compare across drivers/teams.

Relevant tables: races, drivers, lap_times


Same as before we will merge the relevant dataframes. Then, we group by raceId, driverId, and lap number to calculate each driver's lap time per race and lap number across circuits.

```python
# Merge lap times with drivers to get driver info
df2 = lap_times_df.merge(drivers_df, on='driverId', how='left')

# Merge with races to get race info
df2 = df2.merge(races_df[['raceId', 'year', 'circuitId', 'race_name']], on = 'raceId', how = 'left')

# Merge with circuits for track info 
df2 = df2.merge(circuits_df[['circuitId', 'circuit_name']], on = 'circuitId', how='left')

# Compute the average lap time per race, driver and lap number
lap_evolution_df = df2.groupby(['circuitId', 'driverId', 'lap_number']).agg(
    avg_lap_time_millisec=('lap_millisec', 'mean')
).reset_index()

# Merge driver reference for readability
lap_evolution_df = lap_evolution_df.merge(drivers_df[['driverId', 'driverRef']], on='driverId', how='left')

# Merge circuit names for better interpretation
lap_evolution_df = lap_evolution_df.merge(circuits_df[['circuitId', 'circuit_name']], on='circuitId', how='left')

# Sort by race, driver, and lap number
lap_evolution_df = lap_evolution_df.sort_values(by=['circuitId', 'driverId', 'lap_number'])

# Display the result
print(lap_evolution_df.head)
```

```python
lap_evolution_df.to_csv('../../data/avg_laptime_evolution.csv', index=False)
```
